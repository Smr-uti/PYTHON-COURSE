## Global Namespace:

```python
language='Python'

version=3.11
print(language)

def show():
    print(version)
show()
```

```python
player     = 'Bob'
high_score = 4200
level      = 7

def play_level():
    bonus = 500          # LOCAL variable
   # global high_score
    high_score=0
    high_score += bonus  # modify global
    print(player)        # reads global
    print(high_score)

play_level()
```

## Shadowing:

```python
count = 100        # global
def show_count():
   count = 999    # LOCAL shadow!
   print(count)   # → 999
   
show_count()
print(count) 
```

## global keyword:

```python
count = 100        # global
def update_count():
    global count   # ← tell Python!
    count = 999    # modifies global
    print("inside",count)   # → 999
update_count()
print("outside",count) # → 999
```

```python
city       = 'Paris'
population = 2_200_000
country    = 'France'

def city_info():
    district = 'Marais'    # local variable
    global population
    population += 50_000   # modify global
    print('Inside locals()  :', locals())
    print("===========")
    print('Inside globals() :', globals())

# Module level — same dict
print('Module locals() :', locals())
print('Module globals() :', globals())
city_info()
```

```python
jan_sales = 1500
feb_sales = 2200
mar_sales = 1800

months = ['jan_sales', 'feb_sales', 'mar_sales']

for m in months:
    print(m, '=', globals()[m])   # string → variable!

print(globals())
```

```python
def add(x, y):
    return x+y

def multiply(x, y):
    return x*y

def substract(x, y):
    return x-y

operations=['add', 'multiply', 'substract']

for op in operations:
    result= globals()[op](10,3)
    print(f'{op}(10,3)= {result}')
```
