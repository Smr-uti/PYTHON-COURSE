## For loop

```python
square=[]

for n in range(1,6):
    square.append(n**2)

print(square)
```

#### for vs comprehension:

#### Comprehension for square:

```python
square=[n**2 for n in range(1,6)]
print(square)
```

#### Comprehension for cube:

```python
cube=[n**3 for n in range(1,6)]
print(cube)
```

#### Comprehension for even number from 1 to 20 return square:

```python
even_square=[n**2 for n in range(1,21) if n%2==0]
print(even_square)
```

#### Convert word in to uppercase:

```python
animal=["cat", "dog", "fish", "bird"]
upper=[i.upper() for i in animal if len(i)>3]

print(upper)
```

## 1. Set Comprehension:

```python
square_set={n**2 for n in range(1,6)}
print(square_set)

print(type(square_set))
```

#### Remove duplicate letter from word:

```python
text="misssppoooiiiiaaaa"
unq_set={i for i in text}

print(unq_set)
```

## 2. Dictionary Comprehension:

```python
square_dict={n:n**2 for n in range(1,6)}
print(square_dict)
```

```python
animal1=["cat", "dog", "fish", "bird"]
len_animal_dict={i : len(i) for i in animal1}

print(len_animal_dict)
```

```python
students={"akshay":59, "rahul":78, "rajesh":73}

student_more_than_70={k:v for k, v in students.items() if v>70}
print(student_more_than_70)
```
