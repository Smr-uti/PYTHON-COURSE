## Dictionary:

#### A dictionary is a collection of key-value pairs. Keys and values are seprated by comma and enclosed in curly brackets.

```python
dict1={"name":"Akshay", "roll_no": 100, "weight":65.6, "placement": True}
print(dict1)

print(type(dict1))
```

## Features:

#### 1.Keys are unique.

```python
dict1={"name":"Akshay", "roll_no": 100, "weight":65.6, "placement": True, "name":"Rahul"}
print(dict1)
```

#### 2. Different keys, same values.

```python
dict1={"name":"Akshay", "roll_no": 100, "weight":65.6, "placement": True, "name":"Rahul", "pin_code":100}
print(dict1)
```

#### 3. Keys are unique. Each key must be unique--- duplicate keys overwrite.

```python
dict2={"name":"Akshay", "roll_no": 100, "weight":65.6, "placement": True, "name":"Rahul"}
print(dict2)
```

```python
dict1={"name":"Akshay","roll_no":100,"weight":65.6,"placement":True,"name":"Rahul","pin_code":100}
print(dict1)
```

```python
dict1={"name":"Akshay","roll_no":100,"weight":65.6,"placement":True,"Name":"Rahul","pin_code":100}
print(dict1)
```

##### 2. Accessing Dictionary Element

#### 1. Print Entire Dictionary

```python
dict1={"name":"Akshay", "roll_no": 100, "weight":65.6, "placement": True}
print(dict1)
```

#### 2. Accessing value using key. Use the key inside square brackets to retrieve the corresponing value.

```python
dict2={"name":"Akshay", "roll_no": 100, "weight":65.6, "placement": True}
print(dict2)

print(dict1["name"])
print(dict1["roll_no"])
print(dict1["weight"])
print(dict1["placement"])
```

#### 3. No indexing or slicing: Disctionaries do not support integer indexing. You access elements uisng keys, not positions.

## Looping in Dictionaries:

#### 1. Iterate over key value pairs.

```python
list1={10,20,30}
for i in list1:
    print(i)    
```

```python
dict1={"name":"Akshay", "roll_no": 100, "weight":65.6, "placement": True}

for i, j in dict1.items():
    print("Key is:", i, "and Value is:", j)
```

```python
for i in dict1.keys():
    print(i)
```

```python
for j in dict1.values():
    print(j)
```

#### 2.Enumerate: Use enumerate() to get both the iteration index & key-value pairs during looping.

```python
list1=[10,20,30]
for i in list1:
    print(i)    
```

```python
for i, j in enumerate(list1):
    print(i,j)
```

```python
dict1={"name":"Akshay", "roll_no": 100, "weight":65.6, "placement": True}
for i, (k, v) in enumerate(dict1.items()):
    print(i,k,v)
```
