## Set Properties:

#### 1. Mutability: Sets are mutable. After creating set you can add or remove element().

```python
set1={"Smita", "Akshay", "Pooja", "Rohit"}
set1.add("Shivani")
print(set1)
```

```python
set2={"Smita", "Akshay", "Pooja", "Rohit"}
set2.remove("Rohit")
print(set2)
```

```python
set3={"Smita", "Akshay", "Pooja", "Rohit"}
set3.update("Monika")
print(set3)
```

#### 2. Converting other types to set:

#### i]. string-> set

```python
name="maharashtra"
set5=set(name)
print(set5)
```

#### ii]. tuple-> set

```python
tupl=(1,2,3)
set6=set(tupl)
print(set6)
```

#### iii]. list-> set

```python
list1=[1,2,3]
set1=set(list1) 
print(set1)
```

#### iv]. dictionary->set

```python
dict1={"name":"Akshay", "age":25}
set1=set(dict1)
print(set1) 
```

#### v]. range->set

```python
range1=range(10,20)
set2=set(range1)
print(set2)
```

#### 3. Aliasing: Multiple variables point to same memory location.

```python
set1={1,2,3}
set2=set1
print(id(set1))
print(id(set2)) 
```

#### 4. Clonning: copy() It creates copy of set at different memory location.

```python
set={10,20,30}
set2=set1.copy()
print(id(set1))
print(id(set2))
```

#### 5. Searching: 'in' membership operator

```python
set1={10,20,30}
print(30 in set1)
print(50 in set1)
```

#### 6. Identity:'is' operator

```python
set1={10,20,30,40,50}
set2=set1
print(set1 is set2)

print(id(set1))
print(id(set2))
```

#### 7. Comparison: 

```python
set1={10,20,30,40}
set2={10,20,30,35}

print(set1<set2)
print(set1>set2)
```

#### 8. Emptiness: 'not' operator

```python
my_set= set()   
print(my_set)

print(type(my_set))

if not my_set:
    print("set is empty")
else:
    print("set is not empty")
```
