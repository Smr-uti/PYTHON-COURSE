##### Create a decorator for a function called welcome().The decorator should:Print "Welcome message is starting" before calling welcome().Execute the welcome() function.Print "Welcome message completed" after calling welcome()The welcome() function should print:"Welcome to Python"

```python
def welcome(update_message):
    def wrapper():
        update_message()
        print("Welcome to Python")
        print("Welcome Message Completed")
    return wrapper

def message():
    print("Welcome message is Starting")

wrapper = welcome(message)
wrapper()
```

#### Create a decorator for a function called login().The decorator should: Print "Checking login..." before calling login().Execute the login() function. Print "Login process completed" after calling login().

```python
def decor(loggedin):
    def wrapper():
        print("Checking login...")
        loggedin()
        print("Login Process Completed")
    return wrapper

def login():
    print("User logged in successfully !")

wrapper = decor(login)

wrapper()
```

#### Create a decorator for a function called calculate().The decorator should:Print "Calculation Started" before executing calculate(). Execute the calculate() function. Print "Calculation Completed" after executing calculate(). The calculate() function should print:"Calculating two numbers..."

```python
def decor(calculation):
    def wrapper():
        calculation()
        print("Calculation Started")
        print("Calculation Completed...!")
    return wrapper

def calculate():
    print("Calculating two number..:")

wrapper = decor(calculate)

wrapper()

```

#### Create a decorator for a function called greet(name).The decorator should:Print "Greeting Started" before calling greet().Call the original greet(name) function.Print "Greeting Completed" after calling greet().The greet(name) function should print:"Hello, <name>!"

```python
def decor(greeting):
    def wrapper():
        print("Greeting Started...")
        greeting()
        print("Greeting Completed...!")
       
    return wrapper

def greet():
    name = "Smruti"
    print("Hello ", name)
wrapper = decor(greet)
wrapper()
```
