## Regular Function:

```python
def square(n):
    return n**2
result=square(5)
print(result)
```

## 1. lambda function:

```python
square=lambda n:n**2
print(square(5))
```

#### Check if number is positive:

```python
def check_num(n):
    if n>0:
        return True
    else: 
        return False
    
result=check_num(-3)
print(result)
```

```python
check_number= lambda n :n>0
print(check_number(-3))
```

#### Concatenate two strings:

```python
def concat_string(a, b):
    return a+" "+b
result=concat_string("Hello", "Python")
print(result)
```

```python
concat_string=lambda a,b: a+" "+b
print(concat_string("Hello", "Python"))
```

## 2. map fucntion:

#### i]. cube

```python
values=[2,3,4,5,6]
def cube(n): 

    return n**3
result=map(cube, values)

final_result=list(result)

print(final_result)
```

## ii]. price multiple 2:

```python
prices=[10,25,40,55]
double=list(map(lambda p:p*2,prices))

print(double)
```

#### iii] get strings length:

```python
fruits=["apple", "banana", "mango"]
lengths=list(map(lambda l:len(l), fruits))

print(lengths)
```

#### iv] add 18% GST on cost:

```python
costs=[100,200,300]
with_tax=list(map(lambda i:i*1.18, costs))

print(with_tax)
```

## 3. filter() function:

#### i]. filter for marks: jyachi value greater than 50 asel.

```python
marks=[42,48,78,55,33,91,60]
result=filter(lambda s:s>=50,marks)
print(list(result))
```

#### ii]. length greater than 5

```python
fruits=["apple","banana", "mango"]
result=filter(lambda a:len(a)>5, fruits)

print(result)

final_result=list(result)
print(final_result)
```

#### iii]. adults only (age>=18)

```python
ages=[12,25,17,30,15,22]
adults=list(filter(lambda s:s>=18, ages))

print(adults)
```

## 4. reduce():

```python
from functools  import reduce
addition=[1,2,3,4,5]
result=reduce(lambda a, b:a+b, addition)
print(result)
```

#### i]. build a sentence from words:

```python
from functools import reduce
words=["Python","is","awesome"]

sentence=reduce(lambda a,b:a+" "+b,words)
print(sentence)
```
