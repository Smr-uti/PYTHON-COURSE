## Dictionary Methods:

#### 1. update() Method​

```python
my_dict={'a': 1}

my_dict2={'b': 2, 'a': 3}
```

```python
my_dict.update(my_dict2)
```

```python
print(my_dict)
```

```python
print(my_dict["b"])

my_dict["b"]=4
print(my_dict)
```

#### 2. pop(key) Method​:

```python
data = {40: 'a', 30: 'b', 60: 'c'}
value=data.pop(30)
print(value)

print(data)
```

#### 3. popitem()

```python
data = {40: 'a', 30: 'b', 60: 'c'}
data.popitem()

print(data)
```

#### 4. clear() — Remove ALL key-value pairs, making dictionary empty​


```python
data = {40: 'a', 30: 'b', 60: 'c'}
data.clear()
print(data)
```
