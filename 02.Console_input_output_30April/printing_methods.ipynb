#### Accepting two values and returns its addition: Problem

```python
num1=input("Enter first value:")
num2=input("Enter second value:")

result=num1+num2
print(result)
```

#### Accepting two values and returns its addition: Solution

```python
num1=input("Enter first value:")
print(num1)

num2=input("Enter second value:")
print(num2)

result=num1+num2
print(result)
```

```python
num1=int(num1)
print(type(num1))

num2=int(num2)
print(type(num2))

result=num1+num2
print(result)
```

#### Accepting two values and returns its addition: Solution 2

```python
num1=int(input("Enter first value:"))

num2=int(input("Enter second value"))

result=num1+num2 
print(result)
```
