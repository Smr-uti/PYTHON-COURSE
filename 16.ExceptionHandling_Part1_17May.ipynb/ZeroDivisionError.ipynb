## ZeroDivisionError: Example

```python
numerator   = 100
denominator = 0

# Example 1: Basic division by zero
print("=== Example 1: Integer Division by Zero ===")
try:
    result = numerator / denominator   # This will raise ZeroDivisionError
    print('Result:', result)
except ZeroDivisionError as err:
    print('Caught:', err)              # Output: division by zero

# Example 2: Modulo by zero is also illegal
print("\n=== Example 2: Modulo by Zero ===")
try:
    remainder = 17 % 0                 # Modulo by zero — also raises the error
except ZeroDivisionError:
    print('Modulo by zero is illegal!')

# Example 3: Safe helper function — best practice!
print("\n=== Example 3: Safe Division Helper Function ===")

def safe_div(a, b):
    """Returns a/b if b is non-zero, else returns the string 'undefined'."""
    return a / b if b != 0 else 'undefined'

print(safe_div(10, 2))   # Valid division → 5.0

print(safe_div(10, 0))   # Zero denominator → 'undefined' (no crash!)
```

## ZeroDivisionError-Prevention Flow

```python
def safe_div_v2(a, b):
    """A more explicit version of safe division using if/else."""
    if b == 0:
        return 'undefined'   # Safe fallback — no crash
    return a / b             # Normal division

# Practical example: calculating average of a list safely
def safe_average(numbers):
    """Calculate the average of a list, handling empty list."""
    total = sum(numbers)
    count = len(numbers)
    return total / count if count != 0 else 0   # Guard against empty list

print("Average of [10, 20, 30]:", safe_average([10, 20, 30]))  # 20.0
print("Average of []:", safe_average([]))                       # 0 (no crash)
```
