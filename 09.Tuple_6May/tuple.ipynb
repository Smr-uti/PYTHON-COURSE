## Homogeneous Tuple

```python
tup1=("apple","mango","banana")
print(type(tup1))
```

## Heterogenous Tuple

```python
tup2=("Akshay", 100, 65.6, True)
print(type(tup2))
```

## Ordered

```python
tup2=("Akshay", 100, 65.6, True)

print(tup2)

print(tup2[2])
```

## Immutable- no change

```python
tup2=("Akshay", 100, 65.6, True)
```

```python
tup2.append(101)
```

## Duplicates

```python
tup2=("Akshay", 100, 65.6, True, 100, 65.6, 100)

print(tup2)
```

## Empty Tuple

```python
tup3=()

print(tup3)
```

## Single item tuple

```python
tup4=(10)

print(tup4)

print(type(tup4))
```

```python
tup4=(10,)

print(tup4)
print(type(tup4))
```

## Accessing tuple elements

```python
tup1=(10,20,30,40,50)

print(tup1)
```

## Indexing

```python
tup1=(10,20,30,40,50)

print(tup1[0])

print(tup1[2])

print(tup1[-3])
```

## Slicing

```python
tup1=(10,20,30,40,50)

print(tup1[0:3])

print(tup1[1:4])

print(tup1[2:5])

print(tup1[2:])

print(tup1[:3])
```

```python
print(tup1[::2])
```

## Reversing

```python
tup1=(10,20,30,40,50)

print(tup1[::-1])
```

## Looping in Tuple

### 1.while loop

```python
my_tuple=(10,20,30)
index=0

while (index<len(my_tuple)):
    print(my_tuple[index])
    index=index+1
```

## 2.for loop

```python
my_tuple = (10,20,30)

for i in my_tuple:
    print(i)             
```

## 3.enumerate

```python
my_tuple=(10,20,30)

for i, j in enumerate(my_tuple):

         print("Value is:",j,"and Index is:",i)    
```
