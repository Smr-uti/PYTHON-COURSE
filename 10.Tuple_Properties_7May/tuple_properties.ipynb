#### 2. Concatenation:-

```python
tup1=(1,2,3)
tup2=(4,5,6)
tup3=tup1+tup2

print(tup3)
```

#### 3. Merging:-- Combining any numbers of tuple

```python
tup1=(1,2,3)
tup2=(4,5,6)
tup3=(7,8,9)

tup4=tup1+tup2+tup3
print(tup4)
```

#### 4.Converting other types to tuple:
##### i].Converting string to tuple:-- each individual element seprate

```python
word="hello world"
char_tuple=tuple(word)
print(char_tuple)   

print(char_tuple[0])    

print(char_tuple[0:4])
```

##### ii].Converting list to tuple:-- all list element into an ordered immutable tuple.

```python
list1=[1,2,3]
tup1=tuple(list1)
print(tup1)
```

##### iv]. Converting set to tuple:-- converts all its element in tuple. Set are unordered.

```python
set1={1,2,3}
tup1=tuple(set1)

print(set1)

print(tup1)
```

##### v]. Converting dictionary to tuple:-- dictionary jewa tuple mdhe convert hote tewa keys fkt store krte. values ignore krte. (name=key  smruti=value)

```python
dict1={"name":"Smruti","age":30} 
print(dict1)

print(type(dict1))

tup1=tuple(dict1)
print(tup1)

```

##### vi]. Converting range into tuple:-- convert range object into tuple to get immutable sequence of numbers stored as tuple.

```python
num_range=range(1,11,1)
print(type(num_range))

tup1=tuple(num_range)
print(tup1)
print(type(tup1))
```

#### 5. Aliasing:-- ekach memory location la 2 tuple point krnar

```python
tup1=(10,20,30)
print(tup1)
print(type(tup1))

print(id(tup1))
```

```python
tup2=(10,20,30)
print(id(tup2)) 

tup1=tup2
print(tup2)

print(id(tup1))
print(id(tup2)) 
```

##### 6. Clonning:-- creating an independent copy with different memory address

```python
tup1=(10,20,30)
print(type(tup1))
print(id(tup1))

tup2=tup1.deepcopy()

```

##### 7. Searching:-- 'in' membership operator-used to check whether a particular element is present in tuple or not.

```python
nums=(1,2,3,4,5, "Akshay", 65.8)
print(3 in nums)

print("Akshay" in nums) 

print(10 in nums)
```

##### 8. Identity:-- 'is' operator- check if two variable refer to the same memory location

```python
tup1=(1,2,3)
tup2=(4,5,6)

tup1=tup2

print(tup1 is tup2)

print(id(tup1))
print(id(tup2))
```

##### 9. Emptiness:-- 'not' operator- check if tuple is empty

```python
tup1=()
if not tup1:
    print("Tuple is empty")
else:
    print("Tuple is not empty") 
```

```python
tup2=(10,)
if not tup2:
    print("Tuple is empty")
else:
    print("Tuple is not empty") 
```
