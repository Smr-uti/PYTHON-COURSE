## While Loop

```python
print("Program Starts")
list3=[10,20,30,40,50]

index=0
while(index<len(list3)):
    print(list3[index])
    index=index+1

print("Program Ends")
```

## For Loop

```python
print("Program Starts")
list4=[10,20,30,40,50]

for i in list4:
    print(i)

print("Program Ends")
```

## enumerate() Function
Jewa apyala index ani element value donhi print kraych asel tewa enumerate chs use krtat

```python
list5=[10,20,30,40,50]
print("Program Starts")

for index,value in enumerate(list5):
    print("Index is:",index,"and its Value is:",value)

print("Program Ends")
```

## break use krun 
break program terminate krto

```python
print("Program Starts")
list4=[10,20,30,40,50]

for i in list4:
    if i==40:
        break
    print(i)

print("Program Ends")
```

## continue

continue fkt current iteration skip krto. next iteration la jaun next value gheto

```python
print("Program Starts")
list4=[10,20,30,40,50]

for i in list4:
    if i==40:
        continue
    print(i)

print("Program Ends")
```
