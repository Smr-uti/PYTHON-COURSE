## Basic operations:

#### 1. Mutability: Dictionaries are mutable(changeable). After creating dictionary, you can add new key-value pairs, modify existing values, or delete pairs.

```python
dict1={"name":"Akshay", "roll_no":10, "weight":65.6, "placement":True}

dict1={"name":"Ajay"}

print(dict1)
```

#### 2. Concatenation: When two dictionaries are merged using concatenation, they are joined into a single dictionary.


```python
dict1={"name":"Akshay", "roll_no":100}

dict2={"weight":65.6, "placement":True}

dict3={"address":"pune"}

dict4=dict1 | dict2 | dict3

print(dict4)
```

## 3. Converting to Dictionary:

#### i) Convert Tuple of (key,value) Pairs to Dictionary

```python
pairs=(("Name","Akshay"),("Roll_no",100),("Address","Pune"))
print(type(pairs[0]))


print(pairs[0])

dict1=dict(pairs)

print(dict1)
```

```python
pairs[0]
```

#### ii] Convert List of Tuple to Dictionary:

```python
pair=[("Name","Akshay"),("Roll_no",100),("Address","Pune")]
print(type(pair))

dict1=dict(pair)
print(dict1)
```

#### iii] Convert Set elements as key with default values

```python
set1={"a","b","c"}

dict1=dict()

for key in set1:
    print(key)
    dict1[key]=2

print(dict1)    
```

#### iv] Convert Range as Keys with default values:

```python
range1=range(0,5)
dict1={}

for key in range1:
     dict1[key]=10

print(dict1)
```

```python
dict1={}

for i in "hello":
    dict1[i]=10

print(dict1)
```

#### 4. Aliasing:

```python
dict1={"Name":"Akshay", "roll_no":101}

dict2=dict1

print(id(dict2))
```

```python
dict1["Name"]="Rahul" 
print(dict2)
```

#### 5. Clonning:


```python
dict1={"Name":"Akshay", "roll_no":101}
dict2=dict1.copy()

print(id(dict1))

print(id(dict2))    
```

#### 6. Searching -"in" Membership Operator.

```python
dict1={"name":"Akshay", "roll_no":100, "weight":65.6, "placement":True}
print("name" in dict1)

print("address" in dict1)
```

#### 7. 'is' Opearator:

```python
dict1={"name":"Akshay", "roll_no":100, "weight":65.6, "placement":True}
dict2=dict1

print(dict1 is dict2)
```
