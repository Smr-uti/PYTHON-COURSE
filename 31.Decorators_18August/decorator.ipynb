```python
def hello():
    print("Hello")

def my_decorator(func):
    def wrapper():
        print("Good Morning")
        func()
    return wrapper

hello = my_decorator(hello)

hello()
```

```python
def dashboard():
    print("Welcome to Dashboard")

def login_check(func):
    def wrapper():
        print("Checking login...")
        func()
    return wrapper
dashboard = login_check(dashboard)
dashboard()
```

```python
def study():
    print("I am studying")

def greet(action):
    def wrapper():
        print("Welcome Smruti...!")
        action()
    return wrapper()
study = greet(study)
study()
```

```python
def sleep():
    print("I am sleeping")


def night_message(func):
    def wrapper():
        print("Good night..")
        func()

    return wrapper
sleep = night_message(sleep)

sleep()
```
