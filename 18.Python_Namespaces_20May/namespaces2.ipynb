## Inner Functions=Function in Function=Nested Function.

```python
def order_summary():
    item = 'Coffee'
    price = 4.50
    
    print("--- Order Summary ---")

    def show_receipt():
        tax= price* 0.10
        total = price + tax
        print(f'Item: {item}')
        print(f'Price: ${price:.2f}')
        print(f'Tax: ${tax:.2f}')
        print(f'Total: ${total:.2f}')
    show_receipt()
    
order_summary()
```

## LEGB: L- Local | E- Enclosed | G- Global | B- Built-in

```python
x = "Global Variable"  # G - ग्लोबल लेव्हल

def outer_function():
    x = "Enclosing Variable"  # E - एनक्लोजिंग लेव्हल
    
    def inner_function():
        x = "Local Variable"  # L - लोकल लेव्हल
        print(x)
        
    inner_function()

outer_function()
```

```python
country = 'Japan'   # GLOBAL (G)

def outer_fn():
    city = 'Tokyo'  # ENCLOSING (E)
    def inner_fn():
        district = 'Shibuya'   # LOCAL (L)
        # LEGB lookup trace inside inner_fn:
        # district → L (found!) → use it
        # city     → L (not found) → E (found!) → use it
        # country  → L (no) → E (no) → G (found!) → use it
        # print    → L (no) → E (no) → G (no) → B (found!)
        print(district, city, country)
        print(len(district))  # len is BUILT-IN (B)

    inner_fn()

outer_fn()
```

```python
# (B) Built-in: min, max, print — always available
# (G) Global scope
base_wage = 15.0         # GLOBAL

def payroll(hours):      # outer function
    # (E) Enclosing scope
    overtime_rate = 1.5  # ENCLOSING

    def compute_pay():   # inner function
        # (L) Local scope
        regular  = min(hours, 40) * base_wage    # B+E+G
        overtime = max(hours - 40, 0)            # B
        ot_pay   = overtime * base_wage * overtime_rate  # G+E
        total    = regular + ot_pay              # L

        # scope annotation:
        # L: regular, overtime, ot_pay, total
        # E: overtime_rate
        # G: base_wage, hours
        # B: min, max, print
        print(f'Hours:    {hours}')
        print(f'Regular:  ${regular:.2f}')
        print(f'OT Pay:   ${ot_pay:.2f}')
        print(f'Total:    ${total:.2f}')

    compute_pay()

payroll(50)
```

```python
def outer():
    score = 45           # outer's score
    print('outer score:', score)
    print('outer id:   ', id(score))

    def inner():
        score = 90       # inner's OWN score — different!
        print('inner score:', score)
        print('inner id:   ', id(score))

    inner()

outer()
```
