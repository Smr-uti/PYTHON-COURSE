#### Thread Names:

```python
from threading import Thread
import threading
def display():
    msg = "Jay Ganesh"
    print(msg)
```

```python
t1=Thread(target=display)
t1.start()
```

```python
print(t1.name)
```

#### How to change thread names:

```python
t7=Thread(target=display, name="myThread")
t7.start()
t7.name
```
