## KeyError Example

```python
profile = {
    'name':  'Priya',
    'city':  'Pune',
    'score': 94,
}    # Note: 'email' key does NOT exist in this dictionary

# Example 1: Accessing a missing key
print("=== Example 1: Missing Key ===")
try:
    print(profile['email'])      # 'email' is not in the dict
except KeyError as err:
    print('KeyError — missing key:', err)    # Output: 'email'
```

#### Example 2: Case-sensitive mismatch


```python
print("\n=== Example 2: Case Sensitivity ===")
try:
    print(profile['Name'])       # 'name' exists but 'Name' (capital N) does not!
except KeyError as err:
    print('KeyError (wrong case):', err)
```

```python
#Fix 1: .get() — safest way, returns None if key is missing ──
print("\n=== Fix 1: .get() Returns None ===")
email = profile.get('email')     # No KeyError — returns None
print('get():', email)           # None
```

```python
# ── Fix 2: .get() with a default value ──
print("\n=== Fix 2: .get() with Default Value ===")
email = profile.get('email', 'not provided')   # Returns 'not provided' if missing
print('Default:', email)         # not provided
```

```python
# ── Fix 3: 'in' check before accessing ──
print("\n=== Fix 3: 'in' Existence Check ===")
if 'email' in profile:
    print(profile['email'])
else:
    print('No email on record')
```

## KeyError — Prevention Best Practices 

```python
from collections import defaultdict

# Best Practice 1: Always prefer .get() over direct [] access
print("=== Practice 1: .get() Comparison ===")
profile = {'name': 'Priya', 'city': 'Pune', 'score': 94}

# Risky — KeyError if key missing:
# val = profile['email']           ← Avoid this pattern

# Safe — always works:
val = profile.get('email', 'N/A')   # Returns 'N/A' if 'email' not found
print('Safe .get():', val)
```
