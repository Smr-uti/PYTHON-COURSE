#### 2. format() method:

```python
roll_no=100
address="Pune"

print("My Address is: {}, and My Roll No. is: {}" .format(address, roll_no))
```

```python
name="Rahul"
age=25

print("Name:{}, Age:{}".format(name,age))
```

#### 3. fstring() method:

```python
roll_no=100
address="pune"
pin_code=412105
print(f"My Address is: {address}, and Roll No. is {roll_no}, and Pin Code is {pin_code}")
```
