## List Methods

#### 1] Adding Elements

#### 1.append()
end la jr element add krayche asel tr append use krtat

```python
list1=[10,20,30]

print(list1)

print(id(list1))

```

```python
list1.append(40)

print(list1)
```

#### 2. insert()
particular index position la kahi element add kraych asel tr. ani existing element right la shift krto.


```python
list1=[10,20,30,40,50]

list1.insert(4,45)

print(list1)
```

#### 3. extend()
ek list cha end la dusri list add kraychi asel tr extend method use krtat

```python
list1=[10,20,30]

list2=[40,50,60]

list1.extend(list2)

print(list1)
```

## 2]  Removing Elements

#### 4. remove()

jo element remove kraycha ahe tyachi fkt value provide kraychi.
ekhadi value double asel tr hi method pahili value remove krte.

```python
list1=[10,20,30,40,50,30]

list1.remove(30)

print(list1)
```

#### 5. pop()
last element remove krte

```python
list1=[10,20,30,40,50]

list1.pop()
```

#### 6. pop(index)
jo index deil tyanusar delete hoil

```python
list1=[10,20,30,40,50]

list1.pop(3)
```

#### 7. clear()
sglya value delete krun empty list dete

```python
list1=[10,20,30,40,50]

list1.clear()

print(list1)
```

## Searching & Counting:
 

#### i] index():Its finds position of an item. 

```python
list1=[10,20,30,40,50,10]

position=list1.index(10)

print(position)
```

#### ii]. count()
jewdya vela element ala ahe tyacha count milnar

```python
list1=[10,20,30,10,40,50,60,10]

occurance=list1.count(10)

print(occurance)
```

#### iii]. sort(): 
original list assending order ne sort krel

```python
list1=[50,20,40,10,30]

list1.sort()

print(list1)
```

#### iv]. reverse():
list la reverse krte

```python
list1=[50,20,40,10,30]

list1.reverse()

print(list1)
```

#### v]. copy():
list chhi copy bnwte. same elements chi. fkt different location la.

```python
list1=[10,20,30,40,50]

list2=list1.copy()

print(list1)

print(list2)
```
