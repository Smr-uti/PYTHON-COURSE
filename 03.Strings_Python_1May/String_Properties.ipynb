## String Properties
#### 1. Immutability:-- Once created cannot be modified.

```python
str1="Hello World"
print(str1)

print(type(str1))

print(str1[0])

str1[0]="A"
```

#### 2. Concatenation:-- (+) string la jodnysathi use krtat.

```python
greeting="Hello"
name="Rahul"
message=greeting+name

print(message)

print(type(message))
```

```python
message1=greeting+" "+name
print(message1)
```

```python
message2="Hello"+" "+ "Rahul" 
print(message2)
```

#### 3. Repititon (*):-- multiple time string la repeat kraych asel tr 

```python
word="Hi"
repeated_word=word*5
print(repeated_word)
```

```python
word="Hi "
repeated_word=word*5
print(repeated_word)
```

#### 4. Membership Testing(in/not in):-- substring string mdhe ahe ka check krte


```python
text="Python Programming"
print("Python" in text)

print("Java" in text)

print("python" in text) 
```

#### not in:-- jr substring hi string mdhe nasel tr

```python
text1="Python Programming"
print("python" not in text1) 

print("Java" not in text1)
```
