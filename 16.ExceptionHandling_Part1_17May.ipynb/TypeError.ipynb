## TypeError Example:

#### Example 1: str + int → TypeError

```python
item  = 'Laptop'
price = 999

print("=== Example 1: String + Integer Concatenation ===")
try:
    label = item + price         # Cannot concatenate str and int directly
except TypeError as err:
    print('TypeError:', err)
    label = item + str(price)    # Fix: convert int to str first
    print('Fixed:', label)       # Output: Laptop999
```

#### Example 2: len() on a non-sequence type

```python
print("\n=== Example 2: len() on Non-Sequence ===")
try:
    size = len(42)               # integers have no length!
except TypeError as err:
    print('TypeError:', err)
```

#### Example 3: Calling a function with too many arguments

```python
print("\n=== Example 3: Wrong Number of Arguments ===")

def add(a, b):                   # This function accepts exactly 2 arguments
    return a + b

try:
    result = add(1, 2, 3)        # Passing 3 arguments to a 2-parameter function
except TypeError as err:
    print('TypeError:', err)
```

## TypeError — Prevention Best Practices 

#### Practice 1: isinstance() to check type before operating

```python

print("start program")

def safe_double(value):
    """Doubles a number. Raises a clear error if value is not numeric."""
    if not isinstance(value, (int, float)):
        raise TypeError(f"Expected int or float, got {type(value).__name__}")
    return value * 2

print("=== Best Practice: isinstance() Check ===")
print(safe_double(5))       # Works fine
print(safe_double(3.14))    # Works fine
try:
    print(safe_double('5'))  # Will raise our custom TypeError
except TypeError as err:
    print('TypeError:', err)
```

#### Practice 2: Type hints (documentation + IDE support)

```python

def add_numbers(a: int, b: int) -> int:
    """Adds two integers. Type hints document expected types."""
    return a + b

print("\n=== Best Practice: Type Hints ===")
print(add_numbers(3, 4))    # 7 — type hints are just hints, not enforced
```

#### Practice 3: Explicit conversion

```python
user_input = '42'           # Simulating user input (always a string)
number = int(user_input)    # Convert string to int explicitly
print(f"\nConverted user input: {number} (type: {type(number).__name__})")
```
