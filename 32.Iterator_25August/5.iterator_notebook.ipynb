#### Python code_ For loop

```python
def my_own_for_loop(iterable):
    iterator=iter(iterable)
    while True:
        try:
            print(next(iterator))
        except StopIteration:
            break
```

```python
list1=[1,2,3]
print(list1)
```

```python
my_own_for_loop(list1)
```

```python
range1=range(1, 11)
my_own_for_loop(range1)
```
