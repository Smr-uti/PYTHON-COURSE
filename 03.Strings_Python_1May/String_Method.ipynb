### Search and Replace Method:

#### 1.find():-- it returns lowest index of substring/value/word that we provide as a parameter to the method. if substring not fount it returns -1.

```python
str12="Hello, Welcome to Python Programming"
print(str12.find("Python"))
```

```python
str12="Hello, Wecome to Python Programming"
print(str12.find("Java"))
```

#### 2. replace():-- Replaces old substring with new substring. Old substring la new substring mdhe replace krte.

```python
text1="Hello, Welcome to Python Programming"
new_text=text1.replace("Python", "Java")
print(new_text)
```

## Trim Whitespace Methods:
#### 1. lstrip(): It removes leading left whitespace. Shabdacha survatichi whitespace remove krte.

```python
text1="  Programming in Python  "

print(text1.lstrip())
```

#### 2. rstrip():-- It removes trailing whitespace. Right side cha whitespace remove krto.

```python
text2="   Python is Programming Language   "
print(text2.rstrip())
```

#### 3. strip():-- It removes leading and trailing whitespace. Left ani right donhi side cha whitespace remove krte.

```python
text3="   Hello World   "
print(text3.strip())
```

## Split Methods:

#### 1. split():-- It devides string based on delimiter. Eka String la multiple string mdhe devide krte delimiter cha basis wr.

```python
str1="hello, welcome to india" 
print(str1.split())
```

```python
str1="hello, welcome to india" 
words=str1.split()
print(words)
```

```python
str2="Welcome, to our Home"
words=str2.split(",")
print(words)
```

#### 2. partition():-- It devides string into  3 parts. String 3 parts mdhe devide krte.


```python
str3="Who are you?" 
words=str3.partition("are")
print(words)
```

```python
str4="Python is a great language"
words=str4.partition("a")
print(words)
```

## Case Conversion Methods:


#### 1. lower():-- Converts all character in  lowercase.


```python
str5="joKer Is Funny"
print(str5.lower())
```

#### 2. upper():-- Converts all character in uppercase.

```python
str5="joker Is Funny"
sentence=str5.upper()
print(sentence)
```

#### 3. capitalize():-- It converts first character into uppercase rest lowercase. String mdhil fkt first character capital krte.

```python
str6="joker Is Funny"
words=str6.capitalize()
print(words)
```

```python
str7="I am Smruti, I am from Pune"
words=str7.capitalize()
print(words)
```

#### 4. title():--It converts first character of each word into uppercase.

```python
str8="I am Smruti, I am from Pune"
words=str8.title()
print(words)
```

#### 5. swapcase():-- swaps uppercase<-->lowercase

```python
str9="hello WORLD"
result=str9.swapcase()
print(result)
```
