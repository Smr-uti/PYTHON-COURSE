```python
# ── Demonstration: try / except / else / finally ──────────────────────────────
# This cell shows the complete structure of exception handling.
# Run it twice — once with a valid input and once with 0 to see all blocks.

def demonstrate_try_except(numerator, denominator):
    try:
        # The 'try' block contains code that might raise an error
        result = numerator / denominator
    except ZeroDivisionError as err:
        # 'except' block runs only if the error type matches
        print(f"[EXCEPT] Error caught: {err}")
    else:
        # 'else' block runs only when try block had NO error
        print(f"[ELSE]   Success! Result = {result}")
    finally:
        # 'finally' block ALWAYS runs — great for cleanup
        print("[FINALLY] This always executes.")

print("=== Test 1: Valid Division ===")
demonstrate_try_except(10, 2)

print("\n=== Test 2: Division by Zero ===")
demonstrate_try_except(10, 0)
```
