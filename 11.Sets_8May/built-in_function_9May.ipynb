## Built-in function

#### 1. len()- It returns total count of item in set.

```python
set1={10,20,30}
print(len(set1))
```

#### 2. max()- It returns highest value of set.

```python
set2={10,20,30}
print(max(set2))
```

#### 3. min()- It returns lowest value of set.

```python
set3={10,20,30}
print(min(set3))
```

#### 4. sum()- It returns addition of element.


```python
set4={10,20,30}
print(sum(set4))
```

#### 5. any()- True if any element truthy

```python
set5={10,0,0,False}
print(any(set5))
```

#### 6. all()- True only if all items are truthy


```python
set6={10,20,30,40,50}

print(all(set6))
```

#### 7. sorted()- Returns a new sorted LIST in ascending order.

```python
set7={30,10,40,20,50}
print(sorted(set7))
```
