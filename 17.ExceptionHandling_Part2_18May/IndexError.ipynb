## Example 1: Index beyond list length

```python
scores = [88, 72, 95, 61, 78]    # 5 items → valid indices: 0, 1, 2, 3, 4


print("=== Example 1: Index Out of Range ===")
try:
    print(scores[5])             # Index 5 does not exist (last valid = 4)
except IndexError as err:
    print('IndexError:', err)
```

## Example 2: Negative index too far

```python

print("\n=== Example 2: Negative Index Too Far ===")
try:
    print(scores[-10])           # Only -5 to -1 are valid
except IndexError as err:
    print('IndexError (negative):', err)
```

```python
print("\n=== Example 3: Empty List Access ===")
empty = []
try:
    first = empty[0]             # Empty list has no index 0
except IndexError:
    print('IndexError: cannot access empty list!')
```

## Example 4: Safe access using conditional expression

```python
print("\n=== Example 4: Safe Access Pattern ===")
idx = 5
safe = scores[idx] if idx < len(scores) else 'N/A'   # Guard before accessing
print('Safe access at index 5:', safe)    # N/A — no crash!

idx = 2
safe = scores[idx] if idx < len(scores) else 'N/A'
print('Safe access at index 2:', safe)    # 95
```

## IndexError — Prevention Best Practices

####  Practice 1: Check list is non-empty before accessing first/last element

```python
scores = [88, 72, 95, 61, 78]

print("=== Practice 1: Non-empty Check ===")
if scores:                       # Truthy if list has at least one item
    print('First:', scores[0])
    print('Last: ', scores[-1])  # -1 is always the last item (if non-empty)
```

#### Practice 2: enumerate() — avoid manual indexing entirely

```python
print("\n=== Practice 2: Using enumerate() ===")
for idx, score in enumerate(scores):   # enumerate gives (index, value) pairs
    print(f"  Student {idx + 1}: {score} marks")
```

#### Practice 3: try/except for user-supplied indices

```python
print("\n=== Practice 3: try/except for User Input ===")
def get_score_at(lst, index):
    """Safely returns the score at the given index."""
    try:
        return lst[index]
    except IndexError:
        return f'No score at index {index} (list has {len(lst)} items)'

print(get_score_at(scores, 2))    # 95
print(get_score_at(scores, 10))   # Error message — no crash
```

#### Practice 4: Slicing never raises IndexError

```python
print("\n=== Practice 4: Slicing is Always Safe ===")
print(scores[0:3])    # [88, 72, 95]
print(scores[0:10])   # [88, 72, 95, 61, 78] — no error even if 10 > length
```
