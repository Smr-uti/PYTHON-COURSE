## ValueError:

#### Example 1: Converting a non-numeric string to int

```python
import math

print("=== Example 1: int() with Non-Numeric String ===")
try:
    score = int('eighty')    # Type is str (correct), but 'eighty' is not a number
except ValueError as err:
    print('ValueError:', err)
```

#### Example 2: Square root of a negative number

```python
import math

print("\n=== Example 2: math.sqrt() of Negative Number ===")
try:
    result = math.sqrt(-25)  # Type is int (correct), but negative numbers have no real sqrt
except ValueError as err:
    print('ValueError:', err)
```

#### Example 3: Unpacking mismatch

```python
print("\n=== Example 3: Unpacking Mismatch ===")
coords = '10,20,30'          # String with 3 comma-separated values
try:
    x, y = coords.split(',') # Trying to unpack 3 values into 2 variables
except ValueError as err:
    print('ValueError:', err)
```

## ValueError — Prevention Best Practices 

#### Practice 1: Safe integer conversion helper

```python
def to_int(s):
    """Safely converts a string to int. Returns None if conversion fails."""
    try:
        return int(s)
    except ValueError:
        return None    # Return None instead of crashing

print("=== Safe Integer Conversion ===")
print(to_int('42'))      # 42 — valid conversion
print(to_int('eighty'))  # None — invalid, but no crash
```

#### Practice 2: Validate before converting

```python
def get_user_age(input_string):
    """Validates and converts user-supplied age string."""
    if not input_string.isdigit():
        print(f"'{input_string}' is not a valid age.")
        return None
    return int(input_string)

print("\n=== Input Validation ===")
print(get_user_age('25'))      # 25
print(get_user_age('twenty'))  # None, with message
```

#### Practice 3: Domain check for math operations

```python
def safe_sqrt(x):
    """Returns sqrt(x) for non-negative x, else returns None."""
    if x < 0:
        print(f"Cannot compute sqrt of negative number: {x}")
        return None
    return math.sqrt(x)

print("\n=== Safe Square Root ===")
print(safe_sqrt(25))    # 5.0
print(safe_sqrt(-9))    # None, with message
```

#### Practice 4: Controlled unpacking

```python
print("\n=== Safe Unpacking ===")
coords = '10,20,30'
parts = coords.split(',')
if len(parts) == 2:
    x, y = parts
else:
    print(f"Expected 2 coordinates, got {len(parts)}: {parts}")

print("end program")
```
