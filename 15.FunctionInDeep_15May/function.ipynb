## 1. Positional Arguments:

```python
print("Program Starts")
def greet(name, age):
    print("My name is: ", name)
    print("My age is: ", age)

greet(name="Akshay", age=25)
print("End Program")
```

## 2. Keyword Argument:

```python
print("Program Starts")
def greet(name, age):
    print("My name is: ",name)
    print("My age is: ",age)

greet(name="Akshay", age=25)

print("Program Ends")
```

## 3. Default Arguments:

```python
def greet(name,age=25):
    print("My name is:",name)
    print("My age is:", age)

greet("Ganesh")
```

```python
greet("Akshay", 35)
```

## 4. Variable Length Positional Arguments:

```python
print("Start Program")
def cal_sum(*args):
    return (sum(args))

result=cal_sum(1,2,3,4,5,6)
print(result)

print("End Program")
```

```python

print("start program")
def cal_sum(*args):
    print(type(args))
    val1=args[0]
    val2=args[1]
    val3=args[2]
    val4=args[3]
    val5=args[4]
    val6=args[5]
    print(val1)
    print(val2)
    print(val3)
    print(val4)
    print(val5)

cal_sum(1,2,3,4,5,6)

print("end program")
```

## 5. Variable Length Keyword Arguments:


```python
print("start program")
def print_details(**kwargs):
    print(type(kwargs))
    print(kwargs)
    for key,value in kwargs.items():
        print(key,value)

print_details(name="Ganesh",age=35,city="pune")

print("end program")
```
