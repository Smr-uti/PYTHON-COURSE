## Built in function in Tuple

#### 1. len():

```python
tup1=(10,20,30)
print(len(tup1))
```

#### 2. max():


```python
tup1=(10,20,30)
print(max(tup1))
```

#### 3. min():


```python
tup1=(True,1.4,10,20,30,False)
print(min(tup1))
```

#### 4. sum():

```python
tup1=(True,1.4,10,20,30,False)
print(sum(tup1))
```

#### 5. any(): Ektri value True pahije

```python
tup1=(True,1.4,10,20,30,False)
print(any(tup1))
```

#### 6. all(): The all() function returns True only if all items in the tuple evaluate to True. If even a single element is False or 'falsy' (like 0, "", None, or False), it returns False.

```python
tup1 = (True, 1.4, 10, 20, 30, False)
print(all(tup1))
```

#### 7. sorted(): The sorted() function takes any iterable (like a tuple) and returns a new sorted list. It does not modify the original tuple because tuples cannot be changed.

```python
tup1 = (True, 1.4, 10, 20, 30, False)
print(sorted(tup1))
```

#### 8. reversed(): The reversed() function returns an iterator that accesses the tuple elements in reverse order.

```python
tup1 = (True, 1.4, 10, 20, 30, False)
print(tuple(reversed(tup1)))
```

## Built-in methods():

#### 1. index():


```python
tuple1=(10,20,30,40,50)
print(tuple1.index(30))
```

#### 2. count(): It returns occurance of particular element.

```python
tuple1=(10,20,30,40,50,10,10)
print(tuple1.count(10))
```
