## Set Methods:

#### 1. add()- Add a single element.

```python
data={10,20,30,40,50}
data.add(60)
print(data)
```

#### 2. update- Add elements from another set. 2 set bnwawe lagtat dusrya set mdhil value pahilya set mdhe add hotat.

```python
set1={10,20,30}
set2={40,50}

set1.update(set2)
print(set1) 
```

```python
set1={10,20,30}
set2={40,50}

set3={60,70}

set1.update(set2, set3)
print(set1) 
```

```python
set1={10,20,30}
set2={40,50, 30, "Akshay", 10.1, True, 1+2j}

set3={60,70}    

set1.update(set2, set3)
print(set1) 
```

#### 3. copy()- create independent copy of the set

```python
set1={10,20,30}
set2=set1.copy()
print(set2)     
print(id(set2)) 
```

#### 4. remove()- Remove a specific element

```python
set5={10,20,30,40,50}
set5.remove(30)
print(set5) 
```

#### 5. discard()- Remove element, do not error if not found. Safer- no exception raised.

```python
set6={10,20,30,40,50}
set6.discard(30)
print(set6) 
```

```python
set7={10,20,30,40,50}
set7.discard(30)

print(set7) 
set7.discard(90)
```

#### 6. clear()- remove all elements and returns empty set.

```python
data={10,20,30,40,50}
data.clear()

print(data)
```
