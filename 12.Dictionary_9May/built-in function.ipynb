## Built-in Functions:

#### 1. len()- Count of key-value pairs

```python
dict1={"name":"Akshay","roll_no":100,"weight":65.6,"placement":True}
print(len(dict1))
```

#### 2. max()- Largest KEY in dictionary

```python
data = {10:'a', 20.7:'b', True:'c'}
print(max(data))
```

#### 3. min() — Smallest KEY in dictionary

```python
data = {10:'a', 20.7:'b', True:'c'}
print(min(data))
```

#### 4. any() — True if at least one KEY is truthy

```python
dict1 = {0: 'a', 0: 'b', 20: 'c'}
print(any(dict1))
```

#### 5. all() — True only if ALL keys are truthy

```python
dict1 = {0: 'a', 0: 'b', 20: 'c'}
print(all(dict1))
```

#### 6. sorted() — Returns sorted LIST of keys in ascending order​

```python
data = {40: 'a', 30: 'b', 60: 'c'}
print(sorted(data))
```

#### 7. reversed() — Returns reversed order of keys

```python
data = {40: 'a', 30: 'b', 60: 'c'}
print(list(reversed(data)))
```
