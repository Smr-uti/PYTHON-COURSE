## AttributeError: Examples

```python
# ── AttributeError — Example 1: Wrong Method Name ─────────────────────────────
# A very common mistake for developers coming from JavaScript:
# JavaScript arrays use .push() to add items
# Python lists use .append() — .push() does not exist!

items = ['apple', 'mango', 'banana']

print("=== Example 1: Wrong Method Name (JS habit) ===")
try:
    items.push('cherry')         # ❌ .push() is JavaScript, not Python!
except AttributeError as err:
    print('AttributeError:', err)
    items.append('cherry')       # ✅ Correct Python method
    print('Fixed using .append():', items)
```

## Example 2:

```python
# ── AttributeError — Example 2: Accessing Attribute on None ───────────────────
# A very common real-world bug:
# A function returns None (e.g., user not found in database),
# and then you try to access an attribute on that None value.

def find_user(name):
    """Simulates a database lookup — returns None if user not found."""
    # In a real app, this would query a database
    return None    # Simulating: user 'Alice' was not found

user = find_user('Alice')

print("=== Example 2: Attribute Access on None ===")
try:
    print(user.email)            # None has no .email attribute!
except AttributeError as err:
    print('AttributeError:', err)

# ── Fix: Guard against None before accessing attributes ──
print("\n=== Fix: None Guard ===")
if user is not None:
    print('Email:', user.email)
else:
    print('User not found — cannot access email.')
```

## Examples:

```python
# ── Pattern 1: hasattr() check ────────────────────────────────────────────────
# hasattr(obj, 'attr_name') returns True if the attribute exists, else False
print("=== Pattern 1: hasattr() ===")
print(hasattr(items, 'append'))  # True  — .append() exists on list
print(hasattr(items, 'push'))    # False — .push() does NOT exist on list
```

```python
# ── Pattern 2: getattr() with default ─────────────────────────────────────────
# getattr(obj, 'attr_name', default) — returns default if attribute missing
# No exception raised!
print("\n=== Pattern 2: getattr() with Default ===")

class UserProfile:
    def __init__(self, name):
        self.name = name
        # Note: email is NOT set here

profile = UserProfile('Priya')
email = getattr(profile, 'email', 'not provided')   # Safe — no AttributeError
print('Email:', email)           # Output: not provided
name  = getattr(profile, 'name', 'unknown')
print('Name:', name)             # Output: Priya
```

```python
# ── Pattern 3: dir() for discovery ────────────────────────────────────────────
# dir(obj) lists all attributes and methods — great for exploration!
print("\n=== Pattern 3: dir() — Discover Available Methods ===")
list_methods = [m for m in dir([]) if not m.startswith('_')]  # Filter private methods
print('List methods:', list_methods)
```

```python

# ── Pattern 4: Conditional attribute access ────────────────────────────────────
print("\n=== Pattern 4: Conditional Access ===")
def get_user_email(user_obj):
    """Safely retrieves email from a user object."""
    if user_obj is None:
        return 'user not found'
    return getattr(user_obj, 'email', 'email not set')

print(get_user_email(None))       # user not found
print(get_user_email(profile))    # email not set
```
